# Section A: Concept Application

### S1: .py vs .ipynb

**Question:** You open the Heart Disease dataset in Jupyter Notebook. What is the difference between a .py file and a .ipynb file?

**Answer:**
A `.py` file is a standard Python script that contains plain text source code meant to be executed linearly from top to bottom by the Python interpreter. A `.ipynb` (Interactive Python Notebook) file is a JSON document used by Jupyter environments that allows for an interactive mix of executable code cells, rich markdown text, images, and visual outputs (like plots) directly beneath the code that generated them. This makes `.ipynb` ideal for data exploration, reporting, and EDA tasks, while `.py` is preferred for production deployment and module imports.

### S2: Numerical vs Categorical Features

**Question:** The dataset has columns like age, cholesterol, and sex. What is the difference between a numerical feature and a categorical feature?

**Answer:**
*   **Numerical Feature:** Represents quantitative data that can be measured on a continuous or discrete scale (e.g., `age`, `cholesterol`). Meaningful mathematical operations (like calculating the mean or standard deviation) can be performed on numerical features.
*   **Categorical Feature:** Represents qualitative data that is divided into distinct groups or classes (e.g., `sex`, representing male or female). Mathematical operations on these labels are meaningless, even if they are encoded as integers (e.g., 0 and 1). They are usually handled using techniques like One-Hot Encoding or type-casting to 'category' before modeling.

### S3: Data Type of Target Variable

**Question:** You run type() on the target column and it returns int. Why does the data type of your target variable matter before framing a prediction problem?

**Answer:**
The data type of the target variable dictates the fundamental nature of the machine learning task:
*   If the target represents continuous, measurable quantities (usually `float`), it's a **Regression** problem.
*   If the target represents discrete classes or categories (like 0 for no disease and 1 for presence of disease, often `int` or `string`), it's a **Classification** problem.
Knowing this drives every subsequent decision: the choice of algorithms (e.g., Logistic Regression instead of Linear Regression), loss functions (e.g., Cross-Entropy instead of MSE), and evaluation metrics (e.g., Accuracy/F1-Score instead of RMSE).

### S4: Correlation Value

**Question:** You compute the correlation between age and heart disease presence and get 0.23. What does a correlation value tell you about two variables?

**Answer:**
A correlation value (like the Pearson correlation coefficient) ranges from -1 to 1 and indicates the strength and direction of the linear relationship between two variables.
A value of **0.23** suggests a **weak positive linear correlation**. This implies that as `age` increases, the likelihood of `heart disease` tends to slightly increase as well. However, because the value is closer to 0 than to 1, the relationship is relatively weak, meaning `age` alone is not a strongly definitive linear predictor of heart disease.

### S5: Rule-based vs Data-driven Approach

**Question:** A hospital rule says: if cholesterol > 240, flag as high risk. What is the fundamental difference between a rule-based system and a data-driven approach?

**Answer:**
*   **Rule-based System:** Relies on explicitly programmed, static logic defined by human domain experts (e.g., `if chol > 240: risk = High`). It is rigid, does not scale well to hundreds of variables, and struggles to capture complex, hidden interactions between multiple features.
*   **Data-driven Approach:** Uses machine learning algorithms to automatically discover optimal thresholds, patterns, and relationships from historical data. Instead of relying on a human to set a strict cutoff, the algorithm dynamically learns how combinations of features (e.g., high cholesterol combined with age and max heart rate) best predict the outcome, leading to more robust and nuanced predictions.
